```python
from sklearn.datasets import load_iris
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

"""画出聚类的图像
labels：聚类后的label，从0开始的数字
cents；质心坐标
n_cluster:聚类后簇的数量
color：每一簇的颜色
"""
def draw_result(xtrain,label,cents,title):
    n_clusters = np.unique(label).shape[0]
    color = ["red", "orange", "yellow"]
    plt.figure()
    plt.title(title)
    for i in range(n_clusters):
        cur_data = xtrain[label==i]
        # 绘制散点图
        plt.scatter(cur_data[:,0],cur_data[:,1],c=color[i])
        #使用蓝色星形表示中心点位置
        plt.scatter(cents[i,0],cents[i,1],c='blue',marker="*",s=100)
    return plt

    
```

```python
iris = load_iris()
x = iris.data
# 设定聚类数目是三
km = KMeans(n_clusters=3,max_iter=10,n_init=10,init='k-means++',algorithm='elkan',tol=1e-4,random_state=1)
km.fit(x)
print(f"SSE={km.inertia_}")
draw_result(x,km.labels_,km.cluster_centers_,"kmeans").show()
```
